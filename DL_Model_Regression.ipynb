# Deep Learning Regression Model From Scratch

# House Price Prediction

```python
from keras.datasets import boston_housing
(x_train,y_train),(x_test,y_test)=boston_housing.load_data()
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
# Let us create a Sequential Model
from keras.models import Sequential
from keras.layers import Dense, BatchNormalization

# Creating Sequential model
model = Sequential()

model.add(Dense(13, activation="relu", input_shape=(13,)))
model.add(BatchNormalization())

# let us add hidden layer
model.add(Dense(25,activation="relu"))

# let us add hidden layer
model.add(Dense(25,activation="relu"))

# let us add output layer
model.add(Dense(1))

model.summary()
```

```python
# let us compile the model
model.compile(optimizer="adam",
             loss="mse",metrics=["mae"])
```

```python
# let us fit the model
price=model.fit(x_train,y_train,epochs=15,batch_size=20,verbose=1)
```

```python
# let us evaluate the model
model.evaluate(x_test,y_test)
```

```python
# let us visualize
import matplotlib.pyplot as plt
plt.plot(price.history["loss"])
plt.title("Model loss(adam.relu)")
plt.ylabel("loss")
plt.xlabel("Epoch")
plt.show()
```
